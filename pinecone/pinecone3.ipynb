```python
from pinecone import Pinecone, ServerlessSpec

from dotenv import load_dotenv, find_dotenv
import os
load_dotenv(find_dotenv())
```

```python
pc = Pinecone(api_key=os.getenv("PINECONE"))
```

```python
pc.create_index(name='learn-metadata',
                dimension=3,
                spec=ServerlessSpec(cloud="aws", region="us-east-1"))
```

```python
idx = pc.Index('learn-metadata')
```

```python
idx.upsert([
    ("1",[0.1,0.1,0.1,],{"topic":"subject","year":2020}),
    ("2",[0.2,0.2,0.2,],{"topic":"body","year":2021}),
    ("3",[0.3,0.3,0.3,],{"topic":"others","year":2022}),
    ("4",[0.4,0.4,0.4,],{"topic":"subject"}),
    ("5",[0.5,0.5,0.5,],{"topic":"body"})
])
```

```python
idx.query(vector=[0,0,0.1], 
          top_k=3,
          include_metadata=True,
          include_values=True, 
          filter={
              "topic":{"$eq":"subject"}})
```
