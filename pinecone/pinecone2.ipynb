```python
from pinecone import Pinecone, ServerlessSpec

from dotenv import load_dotenv, find_dotenv
import os
load_dotenv(find_dotenv())
```

```python
pc = Pinecone(api_key=os.getenv("PINECONE"))
```

```python
pc.create_index(name='learn-namespace',
                dimension=3,
                spec=ServerlessSpec(cloud="aws", region="us-east-1"))
```

```python
idx = pc.Index('learn-namespace')
```

```python
import numpy as np
```

```python
email_with_subject = 20
email_with_body = 45
email_with_others = 45
```

```python
vec_sub = np.random.rand(email_with_subject,3).tolist()
```

```python
vec_body = np.random.rand(email_with_body,3).tolist()
vec_other = np.random.rand(email_with_others,3).tolist()
```

```python
ids_subj = map(str,np.arange(email_with_subject).tolist())
ids_body = map(str,np.arange(email_with_body).tolist())
ids_other = map(str,np.arange(email_with_others).tolist())
```

```python
vectors_subj = list(zip(ids_subj, vec_sub))
vectors_body = list(zip(ids_body, vec_body))
vectors_other = list(zip(ids_other, vec_other))
```

```python
vectors_other
```

```python
idx.upsert(vectors_subj,namespace='subject')
```

```python
idx.upsert(vectors_body,namespace='body')
idx.upsert(vectors_other,namespace='others')
```

```python
pc.delete_index('learn-namespace')
```
